**Instructions:** Run each cell sequentially. Visually verify the stepper motor rotates.

**Note:** Stepper motors do not use SCPI commands. The stepper driver typically communicates over serial (Arduino).

Direction `1` increases position and `0` decreases it in the supplied sketch.
Check the current position and travel limits before running movement cells.

## 0. Setup & Connection

```python
from piec.drivers.autodetect import autodetect
```

```python
# Virtual preview (no hardware movement)
stepper = autodetect("VIRTUAL_STEPPER_MOTOR")

# Physical hardware: replace the address and use these lines instead.
# from piec.drivers.stepper_motor.arduino_stepper import Geos_Stepper
# stepper = Geos_Stepper("ASRL3::INSTR")
```

## 1. Identification (`idn`)

```python
idn_response = stepper.idn()
print("IDN:", idn_response)
```

**PASS** if an identification string is returned.

---
## 2. Step Tests (`step`)

### 2.1 Step Forward

```python
# Step 100 steps in the forward direction
stepper.step(steps=100, direction=1)
print("Stepped 100 steps forward.")
```

**PASS** if the motor rotates forward.

### 2.2 Step Backward

```python
# Step 100 steps in the reverse direction
stepper.step(steps=100, direction=0)
print("Stepped 100 steps backward.")
```

**PASS** if the motor rotates backward.

### 2.3 Small Step

```python
# Step 1 step forward
stepper.step(steps=1, direction=1)
print("Stepped 1 step forward.")
```

**PASS** if the motor makes a tiny movement.

### 2.4 Large Step

```python
# Step 200 steps forward
stepper.step(steps=200, direction=1)
print("Stepped 200 steps forward.")
```

**PASS** if the motor rotates a significant amount.

```python
# Return to original position
stepper.step(steps=200, direction=0)
print("Stepped 200 steps backward (return).")
```
